```python
## loading the datasets
import pandas as pd

movies = pd.read_csv(r"C:\Users\shivraj\OneDrive\Desktop\ML PROJECTS\Movie recommendation system\tmdb_5000_movies.csv")
credits = pd.read_csv(r"C:\Users\shivraj\OneDrive\Desktop\ML PROJECTS\Movie recommendation system\tmdb_5000_credits.csv")

print(movies.shape)
print(credits.shape)
```

```python
print("\nMovie columns:")
print(movies.columns.tolist())

print("\nCredits columns:")
print(credits.columns.tolist())
```

```python
### Merging the two datasets

movies = movies.merge(
    credits,
    left_on='id',
    right_on='movie_id'
)

print(movies.shape)
```

```python
movies = movies[
    ['movie_id', 'title_x', 'genres', 'keywords', 'overview', 'cast', 'crew']
]

# Rename title_x to title
movies.rename(columns={'title_x': 'title'}, inplace=True)

print(movies.shape)
movies.head()
```

```python
movies[['genres', 'keywords', 'cast', 'crew']].head(1)
```

```python
## converting json like columns into python data

import ast

movies['genres'] = movies['genres'].apply(ast.literal_eval)
movies['keywords'] = movies['keywords'].apply(ast.literal_eval)
movies['cast'] = movies['cast'].apply(ast.literal_eval)
movies['crew'] = movies['crew'].apply(ast.literal_eval)

movies[['genres', 'keywords', 'cast', 'crew']].head(1)
```

```python
def get_names(data):
    return [item['name'] for item in data]

movies['genres'] = movies['genres'].apply(get_names)

movies[['title', 'genres']].head()
```

```python
movies['keywords'] = movies['keywords'].apply(get_names)

movies[['title', 'keywords']].head()
```

```python
def get_cast_names(data):
    return [item['name'] for item in data[:3]]

movies['cast'] = movies['cast'].apply(get_cast_names)

movies[['title', 'cast']].head()
```

```python
def get_director(data):
    for item in data:
        if item['job'] == 'Director':
            return item['name']
    return ''

movies['director'] = movies['crew'].apply(get_director)

movies[['title', 'director']].head()
```

```python
#creating a movie tag
movies['overview'] = movies['overview'].fillna('')

movies['tags'] = (
    movies['genres'] +
    movies['keywords'] +
    movies['cast'] +
    movies['director'].apply(lambda x: [x]) +
    movies['overview'].apply(lambda x: [x])
)

movies[['title', 'tags']].head()
```

```python
##converting tags from a list to a string
movies['tags'] = movies['tags'].apply(lambda x: ' '.join(x))

movies[['title', 'tags']].head()
```

```python
##using TF-IDF vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')

tfidf_matrix = tfidf.fit_transform(movies['tags'])

print(tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)

print(similarity.shape)
```

```python
print(movies['title'].iloc[0])
print(movies['title'].iloc[1])

print(similarity[0][1])
```

```python
def recommend_movies(movie_title):
    movie_index = movies[movies['title'] == movie_title].index[0]

    similarity_scores = similarity[movie_index]

    similar_movies = sorted(
        list(enumerate(similarity_scores)),
        key=lambda x: x[1],
        reverse=True
    )

    for i in similar_movies[1:6]:
        print(movies.iloc[i[0]]['title'])
```

```python
recommend_movies('Avatar')
```
